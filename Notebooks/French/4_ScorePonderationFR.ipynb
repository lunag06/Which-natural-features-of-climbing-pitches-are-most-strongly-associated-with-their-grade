### 3.2 Association des scores à chaque description de voie

```python
# ================================================================
# 2. NORMALISATION DU DICTIONNAIRE
# ================================================================

scores_dictionnaire = {
    (
        str(element).strip().casefold(),
        str(qualificatif).strip().casefold()
    ): score
    for (element, qualificatif), score
    in dictionnaire_match_elementqualifie_diff.items()
}

# Éléments du dictionnaire ayant une colonne de comptage correspondante
elements = list(dict.fromkeys(
    element
    for element, _ in dictionnaire_match_elementqualifie_diff
    if element in df2.columns and f"nb_{element}" in df2.columns
))

print("Nombre d'éléments à pondérer :", len(elements))

# ================================================================
# 3. CALCUL DU SCORE DE CHAQUE OCCURRENCE
# ================================================================

scores_occurrences = []

MODIFICATEURS = {"peu": 0.5, "moins": 0.5, "assez": 0.75,
                 "plus": 1.25, "très": 1.5, "trop": 1.5}

def score_qualificatif(element, qualificatif, scores_dictionnaire):
    """Cherche l'expression complète, puis l'adjectif seul
    avec application du coefficient du modificateur."""
    element = element.casefold()
    texte = str(qualificatif.get("texte", "")).strip().casefold()
    lemme = str(qualificatif.get("lemme", "")).strip().casefold()

    # 1. Expression exacte, sur le texte brut puis sur le lemme complet.
    #    Indispensable pour les négations : « pas facile », « pas évidente ».
    for cle in [(element, texte), (element, lemme)]:
        if cle in scores_dictionnaire:
            return scores_dictionnaire[cle]

    # 2. Adjectif seul + coefficient du modificateur. On essaie le lemme
    #    ET le texte brut : spaCy lemmatise « soutenue » en « soutenir »,
    #    « couché » en « coucher », donc le lemme seul ne matche pas
    #    les entrées du dictionnaire.
    for source in (lemme, texte):
        mots = source.split()
        if len(mots) > 1 and mots[0] in MODIFICATEURS:
            adj = " ".join(mots[1:])
            base = scores_dictionnaire.get((element, adj))
            if base is not None:
                return float(
                    np.clip(base * MODIFICATEURS[mots[0]], -1.0, 1.0)
                )

    # 3. Adjectif seul, sans modificateur : lemme puis texte brut.
    return scores_dictionnaire.get(
        (element, lemme),
        scores_dictionnaire.get((element, texte), 0)
    )

for _, occurrence in df.iterrows():

    index = occurrence["index_dataframe"]
    element = occurrence["terme_canonique"]

    if pd.isna(index) or pd.isna(element):
        continue

    index = int(index)
    element = str(element).strip()

    if not (0 <= index < len(df2)):
        continue

    if element not in elements:
        continue

    # Lecture des qualificatifs de l'occurrence
    valeur = occurrence["qualificatifs_retenus"]

    if pd.isna(valeur) or not str(valeur).strip():
        qualificatifs = []
    elif isinstance(valeur, list):
        qualificatifs = valeur
    else:
        qualificatifs = json.loads(valeur)

    # Chaque qualificatif reçoit son score.
    # Un qualificatif absent du dictionnaire reçoit 0.
    scores_qualificatifs = []

    for qualificatif in qualificatifs:

        if not str(qualificatif.get("texte", "")).strip():
            continue

        score = score_qualificatif(element, qualificatif, scores_dictionnaire)

        scores_qualificatifs.append(score)

    # Score individuel : moyenne UNIQUEMENT sur les qualificatifs connus
    # du dictionnaire. Les inconnus (bruit ou non scorés) ne diluent plus
    # les vrais scores vers 0.
    scores_connus = [s for s in scores_qualificatifs if s != 0]
    score_occurrence = np.mean(scores_connus) if scores_connus else 0

    scores_occurrences.append({
        "index_dataframe": index,
        "terme_canonique": element,
        "score_occurrence": score_occurrence
    })

# Tableau des scores individuels
df_scores = pd.DataFrame(scores_occurrences)

# ================================================================
# 4. MOYENNE DES OCCURRENCES PAR LONGUEUR ET PAR ÉLÉMENT
# ================================================================

if not df_scores.empty:
    df_scores_nonnuls = df_scores[df_scores["score_occurrence"] != 0]
    moyennes = (
        df_scores_nonnuls
        .groupby(["index_dataframe", "terme_canonique"])["score_occurrence"]
        .mean()
    )

# ================================================================
# 5. AJOUT DES COLONNES AU DATASET FINAL
# ================================================================

df3 = df2.copy()

for element in elements:

    df3[f"{element}_ponderation"] = [
        moyennes.get((index, element), 0)
        for index in range(len(df3))
    ]

# ================================================================
# 6. VÉRIFICATION
# ================================================================

colonnes_ponderation = [
    f"{element}_ponderation"
    for element in elements
]

print("Nombre de colonnes de pondération ajoutées :",
      len(colonnes_ponderation))

display(
    df3[
        ["document_id", "longueur", "nb_dalle", "dalle",
         "dalle_ponderation"]
    ].head(10)
)

print("\nRépartition des scores de dalle :")
df3["dalle_ponderation"].value_counts().sort_index()
```
